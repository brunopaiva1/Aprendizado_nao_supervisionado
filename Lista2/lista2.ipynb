**1)** Dado 02 v.a. $X$ e $Y$ Normais, independentes com média zero e variância 1, gere sequências de 5.000 amostras cada para uma delas. Então:


```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

num_amostras = 5000

X = stats.norm.rvs(loc=0, scale=1, size=num_amostras)
Y = stats.norm.rvs(loc=0, scale=1, size=num_amostras)

print("Variáveis X e Y")
print(f"Média de X: {X.mean():.4f}, Variância de X: {X.var():.4f}")
print(f"Média de Y: {Y.mean():.4f}, Variância de Y: {Y.var():.4f}")
```

* **a.** Obtenha a $\text{Corr}(X,Y)$, a $\text{Cov}(X,Y)$ e plote o gráfico de dispersão para $XY$ (scatter de $X-Y$).

```python
cov_matrix_xy = np.cov(X, Y)
cov_xy = cov_matrix_xy[0, 1]

corr_matrix_xy = np.corrcoef(X, Y)
corr_xy = corr_matrix_xy[0, 1]

print(f"a) Análise de X e Y:")
print(f"   - Cov(X,Y) = {cov_xy:.4f}")
print(f"   - Corr(X,Y) = {corr_xy:.4f}\n")

plt.figure(figsize=(8, 6))
plt.scatter(X, Y, alpha=0.6, edgecolors='k')
plt.title('Gráfico de Dispersão para X e Y')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```

* **b.** Obtenha a $\text{Corr}(W,Z)$, a $\text{Cov}(W,Z)$, o coeficiente de correlação e plote o gráfico de dispersão para $WZ$, sendo $W = 2X + 5$ e $Z = Y$.

```python
W = 2 * X + 5
Z = Y

cov_wz_b = np.cov(W, Z)[0, 1]
corr_wz_b = np.corrcoef(W, Z)[0, 1]

print(f"b) Análise com W = 2X + 5 e Z = Y:")
print(f"   - Cov(W,Z) = {cov_wz_b:.4f}")
print(f"   - Corr(W,Z) = {corr_wz_b:.4f}\n")

plt.figure(figsize=(8, 6))
plt.scatter(W, Z, alpha=0.6, edgecolors='k')
plt.title('Gráfico de Dispersão para W e Z (item b)')
plt.xlabel('W = 2X + 5')
plt.ylabel('Z = Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```

* **e.** Obtenha a $\text{Corr}(W,Z)$, a $\text{Cov}(W,Z)$, o coeficiente de correlação e plote o gráfico de dispersão para $WZ$, sendo $W = 2\cos(\theta)X - \text{sen}(\theta)Y$ e $Z = 2\text{sen}(\theta)X + \cos(\theta)Y$. Para $\theta = 0$.
* **f.** Obtenha a $\text{Corr}(W,Z)$, a $\text{Cov}(W,Z)$, o coeficiente de correlação e plote o gráfico de dispersão para $WZ$, sendo $W = 2\cos(\theta)X - \text{sen}(\theta)Y$ e $Z = 2\text{sen}(\theta)X + \cos(\theta)Y$. Para $\theta = \pi/4$.
* **g.** Obtenha a $\text{Corr}(W,Z)$, a $\text{Cov}(W,Z)$, o coeficiente de correlação e plote o gráfico de dispersão para $WZ$, sendo $W = 2\cos(\theta)X - \text{sen}(\theta)Y$ e $Z = 2\text{sen}(\theta)X + \cos(\theta)Y$. Para $\theta = \pi/2$.
* **h.** Obtenha a $\text{Corr}(W,Z)$, a $\text{Cov}(W,Z)$, o coeficiente de correlação e plote o gráfico de dispersão para $WZ$, sendo $W = 2\cos(\theta)X - \text{sen}(\theta)Y$ e $Z = 2\text{sen}(\theta)X + \cos(\theta)Y$. Para $\theta = 3\pi/4$.

```python
def analisar_rotacao(X, Y, theta, item_letra):

    W = 2 * np.cos(theta) * X - np.sin(theta) * Y
    Z = 2 * np.sin(theta) * X + np.cos(theta) * Y

    cov_wz = np.cov(W, Z)[0, 1]
    corr_wz = np.corrcoef(W, Z)[0, 1]

    print(f"{item_letra}) Análise para theta = {theta/np.pi:.2f}π:")
    print(f"   - Cov(W,Z) = {cov_wz:.4f}")
    print(f"   - Coeficiente de Correlação = {corr_wz:.4f}\n")

    plt.figure(figsize=(8, 6))
    plt.scatter(W, Z, alpha=0.6, edgecolors='k')
    plt.title(f'Gráfico de Dispersão para W e Z (item {item_letra}, θ={theta/np.pi:.2f}π)')
    plt.xlabel('W')
    plt.ylabel('Z')
    plt.grid(True)
    plt.axis('equal')
    plt.show()
```

```python
# e. Para theta = 0
analisar_rotacao(X, Y, 0, 'e')

# f. Para theta = pi/4
analisar_rotacao(X, Y, np.pi/4, 'f')

# g. Para theta = pi/2
analisar_rotacao(X, Y, np.pi/2, 'g')

# h. Para theta = 3*pi/4
analisar_rotacao(X, Y, 3*np.pi/4, 'h')
```

**2)** Considere uma função densidade de probabilidade (fdp) Normal bidimensional $f_{X,Y}(X,Y)$ com médias $m_x = 5$, $m_y = 3$ e matriz de covariância $P = [4\ 1.8;\ 1.8\ 1]$. Gere 100.000 amostras para essa distribuição. Então:

```python
from scipy.stats import multivariate_normal

medias = [5, 3]
matriz_cov = [[4, 1.8], [1.8, 1]]
num_amostras_2 = 100000

dist_multi = multivariate_normal(mean=medias, cov=matriz_cov)
amostras = dist_multi.rvs(size=num_amostras_2)

X_multi = amostras[:, 0]
Y_multi = amostras[:, 1]
```

* **a.** Plote o Gráfico 3D de $f_{X,Y}(X,Y)$.

```python
x_grid, y_grid = np.mgrid[0:10:.1, -1:7:.1]
pos = np.empty(x_grid.shape + (2,))
pos[:, :, 0] = x_grid
pos[:, :, 1] = y_grid
fdp_vals = dist_multi.pdf(pos)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x_grid, y_grid, fdp_vals, cmap='viridis')
ax.set_title('Gráfico 3D da fdp Normal Bidimensional')
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.show()
```

* **b.** Plote as superfícies de níveis de $f_{X,Y}(X,Y)$.

```python
plt.figure(figsize=(8, 6))
plt.contour(x_grid, y_grid, fdp_vals, cmap='viridis')
plt.scatter(X_multi, Y_multi, alpha=0.1)
plt.title('Superfícies de Nível e Dispersão das Amostras')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.axis('equal')
plt.show()
```
